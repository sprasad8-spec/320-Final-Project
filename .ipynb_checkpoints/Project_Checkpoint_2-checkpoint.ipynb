# Bikeshare Usage Based on Weather Intensity

The data gives four categories of weather, which essentially divide themselves into Clear/Low/Medium/High in terms of intensity. What can we say about bikeshare usage given weather intensity on a particular day? We can use either ANOVA or a T-Test in order to look at differences in groups.
- 1: Clear, Few clouds, Partly cloudy, Partly cloudy
- 2: Mist + Cloudy, Mist + Broken clouds, Mist + Few clouds, Mist
- 3: Light Snow, Light Rain + Thunderstorm + Scattered clouds, Light Rain + Scattered clouds
- 4: Heavy Rain + Ice Pallets + Thunderstorm + Mist, Snow + Fog

```python
df_intens1 = df[df["weathersit"] == 1]["cnt"]
df_intens2 = df[df["weathersit"] == 2]["cnt"]
df_intens3 = df[df["weathersit"] == 3]["cnt"]
df_intens4 = df[df["weathersit"] == 4]["cnt"]

df_intensheadcount = pd.DataFrame({
    "Weather Intensity:": ["1", "2", "3", "4"],
    "Count:": [df_intens1.count(), df_intens2.count(), df_intens3.count(), df_intens4.count()]
})

df_intensheadcount
```

We don't have many representatives from Intensity 4 weather, not a lot from 3 either.

```python
print(df_intens1.describe())
print(df_intens2.describe())
print(df_intens3.describe())
print(df_intens4.describe())
```

```python
plt.hist(df_intens1, bins = 30)
plt.hist(df_intens2, bins = 30, alpha = 0.8)
plt.hist(df_intens3, bins = 60, alpha = 0.5)
plt.legend(["Intensity 1", "Intensity 2", "Intensity 3"])
plt.xlabel("Bikeshare Usage on Day")
plt.ylabel("Day Frequency")
plt.title("Histogram of Bikeshare Usage on Given Day, Separated by Weather Intensity")
```

Looking at the summary statistics and histogram, we probably won't get much out of copmaring Intensity 3 to the others, but we should compare Intensities 1-3 anyways.

Let $\alpha = 0.05$

$H_0$: The intensity of weather has no bearing on Bikeshare usage on a given day.

$H_A$: The intensity of weather has a bearing on Bikeshare usage on a given day.

```python
stats.f_oneway(df_intens1, df_intens2, df_intens3)
```

# Are there more registered bikers on working days?

```python
df_workday = df.groupby('workingday')[['casual','registered']].mean()
df_workday.columns = ['Casual Users', 'Registered Users']

fig, ax = plt.subplots(layout='constrained')
df_workday.plot(kind='bar', ax=ax, rot=0)
# Title and other labels

ax.set_ylabel('Users')
ax.set_title('Mean Values of Casual and Registered Bikeshare Users by Working Day')
ax.legend(loc='upper left', ncols=3)
ax.set_xticklabels(['No','Yes'])

plt.show()
```

This bar chart shows that while there are more registered Bikeshare users than casual ones, on average, the number of registered users is significantly higher and the number of casual users is lower on working days. This could mean that work travel is a notable reason for registering, and people tend to use Bikeshare casually in their free time outside of work.

# `Bikeshare Usage Based on Temperature and Feeling Temperature`

In this section, we look at how temperature affects daily bikeshare usage. We compare actual temperature to feeling temperature to see which one is more strongly related to ridership, and we test whether the gap between the two affects how many people ride. We use chi-square tests of independence on grouped data (low/medium/high), and we removed one data-entry error before running them. We found that both temperature measures are strongly related to ridership with almost no difference between them, and that the gap only seems to matter because it grows on warm days, when ridership is already higher.

```python
# convert temp to fahrenheit
df["temp_f"] = df["temperature_fahrenheit"] * 41 * 9/5 + 32

# feeling temperature was already in F
df["atemp_f"] = df["feeling_temperature_fahrenheit"]


# making rides a category: days above the median count are "High", the rest are "Low"
df["rides"] = np.where(df["cnt"] > df["cnt"].median(), "High", "Low")

# run a chi-square test for each of the three variables
for col, name in [("atemp_f", "Feeling temp")]:

    group = np.where(df[col] > df[col].median(), "Above median", "Below median")
    table = pd.crosstab(group, df["rides"])

    # chi-square test of independence: returns the test statistic, p-value,
    # degrees of freedom, and the counts we'd expect if there were no relationship
    chi2, p, dof, expected = chi2_contingency(table)

    print(name, "chi =", round(chi2, 1), "p =", p)
    print(table, "\n")
```

$\alpha = 0.05$

Testing feeling temp vs rides

$H_0$: Warm vs. cool feeling days has no relationship with high vs. low ridership.

$H_A$: Warm vs. cool feeling days is related to ridership.

The 3.0E-42 p-value is far below 0.05, so we reject H₀ each time.



Conclusion: About 75% of warm days are high-ridership days (273 of 364), vs. about 25% of cool days (92 of 366). Feeling temperature has a slightly larger χ² than actual temperature (185.5 vs. 179.5), but the gap between them is tiny, so they’re basically equal. Days with a large gap also tend to be high-ridership days, but the gap is larger on warm days, so this partly reflects temperature.
